## Dropping Unwanted columns in the dataset

```python
netflix_df.drop(['director','cast','date_added'],axis=1,inplace=True)
```

```python
netflix_df.head()
```

```python
netflix_df['country'].isnull().sum()
```

### Replacing Null values in country with united states as many of movies were produced there

```python
netflix_df['country'].replace(np.NaN,'United States',inplace=True)
netflix_df.head()
```

```python
netflix_df['country'].isnull().sum()
```

```python
netflix_df['rating'].isnull().sum()
```

### Replacing Null values in rating with TV-MA

```python
netflix_df['rating'].replace(np.NaN,'TV-MA',inplace=True)
netflix_df
```

```python
netflix_df['rating'].isnull().sum()
```

### Now our data does not have any Null values

```python
netflix_df.isnull().sum()
```

## Plotting Pie plot for the content type in Netflix

```python
plt.figure(figsize=(15,15))
plt.pie(netflix_df['type'].value_counts().sort_values(),labels=netflix_df['type'].value_counts().index,explode=[0.05,0],
        autopct='%.2f%%',colors=['yellow','darkblue'])
plt.show()
```

```python
import warnings
warnings.filterwarnings("ignore")
```

## Rating of Movie and Tv show in Bar Plot

```python
sns.countplot(netflix_df['rating'])
sns.countplot(netflix_df['rating']).set_xticklabels(sns.countplot(netflix_df['rating']).get_xticklabels(), rotation=90, ha="right")
fig = plt.gcf()
fig.set_size_inches(10,10)
plt.title('Rating');
```

## Relation Between Type and Rating of Movies and Tv shows in Bar Plot

```python
plt.figure(figsize=(10,8))
sns.countplot(x='rating',hue='type',data=netflix_df)
plt.title('Relation between Type and Rating of Movies and TV Shows ')
plt.show();
```

## Top 5 Countries with most number of movies

```python
most_movies_produced = netflix_df.groupby('country').count().sort_values('title',ascending=False).head(5)
most_movies_produced.reset_index(inplace=True)
most_movies_produced
```

```python
plt.bar(most_movies_produced.country,most_movies_produced.title);
```

## Top 5 years with number of most releases

```python
top_5_year = netflix_df.groupby('release_year').count().sort_values('title',ascending=False).head(5)
top_5_year.reset_index(inplace=True)
top_5_year
```

```python
plt.bar(top_5_year.release_year,top_5_year.title);
```

```python
plt.figure(figsize=(15,5))
sns.barplot(x = netflix_df["listed_in"].value_counts().head(10).index,
            y = netflix_df["listed_in"].value_counts().head(10).values,palette="BuGn")
plt.xticks(rotation=80)
plt.title("Top10 Genre in Movies",fontweight="bold")
plt.show()
```
